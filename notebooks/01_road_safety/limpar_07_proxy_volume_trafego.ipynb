Dado 7: Volume de Veículos por Via (Adaptado via Hierarquia Funcional)

• Fonte Oficial / Técnica: Referência EPTC Transparente; extração vetorial via OpenStreetMap (highway).
• Justificativa Metodológica: Como os medidores eletrônicos da EPTC cobrem apenas pontos isolados da cidade, utiliza-se a classificação funcional da via (primary, secondary, tertiary, residential) como indicador indireto (proxy) de intensidade de fluxo veicular para toda a malha urbana.
• Entrada: dados_brutos/bruto_osm_malha_poa.geojson
• Saída: dados_limpos/limpo_07_proxy_volume_trafego.geojson

```python
import geopandas as gpd

ruas = gpd.read_file("../../data/1_raw/bruto_osm_malha_poa.geojson")

# Limpa colchetes/aspas e pega apenas o primeiro tipo de via usando partition (sem precisar de colchetes)
texto_limpo = ruas['highway'].astype(str).str.replace(r"[\[\]']", "", regex=True)
ruas['hierarquia_viaria'] = texto_limpo.str.partition(',')[0].str.strip()
ruas['extensao_metros'] = ruas['length'].astype(float).round(2)

colunas = [c for c in ['u', 'v', 'key', 'name', 'hierarquia_viaria', 'extensao_metros', 'geometry'] if c in ruas.columns]
ruas[colunas].to_file("../../data/3_done/limpo_07_proxy_volume_trafego.geojson", driver="GeoJSON")

print(f"Concluído! {len(ruas)} trechos salvos em 'dados_limpos/'.")
```
