
Extração de Dados Brutos: Malha Viária de Porto Alegre (OpenStreetMap)

Objetivo: Coletar a base viária bruta (nós/esquinas e arestas/trechos de rua) do município de Porto Alegre via API Overpass (OSMnx), sem aplicar filtros ou cálculos prévios, garantindo a rastreabilidade dos dados brutos no repositório.

```python
import osmnx as ox

ox.settings.requests_timeout = 300
print("Baixando dados 100% brutos do OpenStreetMap...")

grafo = ox.graph_from_place("Porto Alegre, Rio Grande do Sul, Brazil", network_type="bike")
nos, ruas = ox.graph_to_gdfs(grafo)

for col in nos.columns:
    if col != 'geometry':
        nos[col] = nos[col].astype(str)

for col in ruas.columns:
    if col != 'geometry':
        ruas[col] = ruas[col].astype(str)

nos.to_file("../../data/1_raw/bruto_osm_nos_poa.geojson", driver="GeoJSON")
ruas.to_file("../../data/1_raw/bruto_osm_malha_poa.geojson", driver="GeoJSON")
print("Arquivos brutos salvos com sucesso em 'dados_brutos/'!")
```
