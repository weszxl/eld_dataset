Auditoria de Qualidade, Completude e Impacto nos Dados Tratados

* Objetivo: Inspecionar a integridade das bases originais (`dados_brutos/`), mensurar a completude das variáveis críticas e diagnosticar **quais arquivos tratados (`dados_limpos/`) sofrem impacto** por lacunas na fonte oficial ou colaborativa.
* Dimensões Avaliadas:
  1. Cobertura Temporal e Espacial (EPTC): Período coberto pelos registros de sinistros e percentual de acidentes perdidos no georreferenciamento por ausência de coordenadas (`latitude`/`longitude`) na base da prefeitura.
  2. Completude da Malha Viária (OpenStreetMap): Percentual de preenchimento real de atributos estruturais (`name`, `highway`, `maxspeed`, `length`) e extensão total da rede viária mapeada em Porto Alegre.
  3. Mapeamento de Prejuízos por Arquivo Limpo: Identificação direta de quais outputs (`limpo_01` a `limpo_07`) possuem dados ausentes ou limitações de cobertura, acompanhada da causa raiz e da recomendação técnica para a etapa de modelagem do grupo.

```python
import os
import pandas as pd
import geopandas as gpd

print("=" * 80)
print("AUDITORIA FINAL DE INTEGRIDADE DOS DADOS LIMPOS (PILAR SEGURANÇA VIÁRIA)")
print("=" * 80)

pasta = "../dados_limpos"
for arq in sorted(os.listdir(pasta)):
    caminho = os.path.join(pasta, arq)
    if arq.endswith(".csv"):
        df = pd.read_csv(caminho)
    elif arq.endswith(".geojson"):
        df = gpd.read_file(caminho)
    else:
        continue

    print(f"\n• Arquivo: {arq} | Registros: {len(df)}")
    for col in df.columns:
        if col == 'geometry':
            nulos = df.geometry.is_empty.sum() + df.geometry.isna().sum()
        else:
            nulos = df[col].isna().sum() + df[col].astype(str).isin(['None', 'nan', 'NaN', '']).sum()
        pct = ((len(df) - nulos) / len(df)) * 100 if len(df) > 0 else 0
        print(f"   - {col:<25}: {pct:6.2f}% preenchido ({nulos} vazios)")
```
