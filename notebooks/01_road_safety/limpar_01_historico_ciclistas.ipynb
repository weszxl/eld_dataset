Dado 1: Histórico de Acidentes com Ciclistas

• Fonte Oficial: Portal de Dados Abertos de Porto Alegre (DataPOA / EPTC).
• Entrada: dados_brutos/bruto_eptc_acidentes.csv
• O que este script faz: Filtra a base geral de sinistros da EPTC para manter apenas ocorrências que envolveram bicicletas (bicicleta > 0), padroniza os nomes das colunas e seleciona as informações temporais e de localização descritiva (data, hora, dia da semana e logradouro).
• Saída: dados_limpos/limpo_01_historico_ciclistas.csv

```python
import pandas as pd

df = pd.read_csv("../../data/1_raw/bruto_eptc_acidentes.csv", sep=";", low_memory=False)
df.columns = df.columns.str.lower().str.strip()

# Filtra acidentes envolvendo bicicletas
df_ciclistas = df[df['bicicleta'] > 0].copy()

# Mantém colunas históricas relevantes que existirem na base
colunas_desejadas = ['id', 'data', 'hora', 'dia_sem', 'log1', 'log2', 'predial1', 'bicicleta', 'tipo_acid']
colunas_presentes = [c for c in colunas_desejadas if c in df_ciclistas.columns]

df_limpo = df_ciclistas[colunas_presentes].dropna(how='all')
df_limpo.to_csv("../../data/3_done/limpo_01_historico_ciclistas.csv", index=False)

print(f"Concluído! {len(df_limpo)} registros históricos de ciclistas salvos.")
```
