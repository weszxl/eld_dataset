Dado 2: Quantidade de Acidentes por Gravidade

• Fonte Oficial: Portal de Dados Abertos de Porto Alegre (DataPOA / EPTC).
• Entrada: dados_brutos/bruto_eptc_acidentes.csv
• O que este script faz: Isola os acidentes com ciclistas e estrutura a severidade de cada ocorrência (quantidade de feridos, feridos graves, mortes e criação de um índice numérico ponderado de gravidade para uso futuro no modelo de Machine Learning).
• Saída: dados_limpos/limpo_02_gravidade_acidentes.csv

```python
import pandas as pd

df = pd.read_csv("../../data/1_raw/bruto_eptc_acidentes.csv", sep=";", low_memory=False)
df.columns = df.columns.str.lower().str.strip()

df_ciclistas = df[df['bicicleta'] > 0].copy()

# Garante valores numéricos para feridos e fatais
for col in ['feridos', 'feridos_gr', 'fatais', 'mortes']:
    if col in df_ciclistas.columns:
        df_ciclistas[col] = pd.to_numeric(df_ciclistas[col], errors='coerce').fillna(0).astype(int)

col_fatais = 'fatais' if 'fatais' in df_ciclistas.columns else 'mortes'
df_ciclistas['indice_gravidade'] = (df_ciclistas['feridos'] * 1) + (df_ciclistas[col_fatais] * 5)

colunas_saida = [c for c in ['id', 'data', 'feridos', 'feridos_gr', col_fatais, 'indice_gravidade'] if c in df_ciclistas.columns]
df_ciclistas[colunas_saida].to_csv("../../data/3_done/limpo_02_gravidade_acidentes.csv", index=False)

print("Concluído! Base de gravidade salva em 'dados_limpos/'.")
```
