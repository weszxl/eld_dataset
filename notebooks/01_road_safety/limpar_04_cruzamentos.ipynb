Dado 4: Quantidade de Cruzamentos e Interseções por Via

• Fonte Técnica: OpenStreetMap (via OSMnx).
• Entrada: dados_brutos/bruto_osm_malha_poa.geojson e dados_brutos/bruto_osm_nos_poa.geojson
• O que este script faz: Cruza os nós viários (esquinas) com os trechos de rua para contabilizar quantas vias se conectam nas extremidades de cada segmento, medindo a complexidade de interseções de cada trecho.
• Saída: dados_limpos/limpo_04_cruzamentos.geojson

```python
import geopandas as gpd

nos = gpd.read_file("../../data/1_raw/bruto_osm_nos_poa.geojson")
ruas = gpd.read_file("../../data/1_raw/bruto_osm_malha_poa.geojson")

nos['street_count'] = nos['street_count'].astype(float).astype(int)
mapa_nos = nos.set_index('osmid')['street_count']

ruas['cruzamentos_inicio'] = ruas['u'].astype(int).map(mapa_nos).fillna(0).astype(int)
ruas['cruzamentos_fim'] = ruas['v'].astype(int).map(mapa_nos).fillna(0).astype(int)
ruas['total_intersecoes'] = ruas['cruzamentos_inicio'] + ruas['cruzamentos_fim']

colunas = [c for c in ['u', 'v', 'key', 'name', 'total_intersecoes', 'geometry'] if c in ruas.columns]
ruas[colunas].to_file("../../data/3_done/limpo_04_cruzamentos.geojson", driver="GeoJSON")

print("Concluído! Contagem de interseções salva em 'dados_limpos/'.")
```
